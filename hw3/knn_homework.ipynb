```python
def euclideanDistance(instance1, instance2, length):
   distance = 0
   for x in range(length):
         #print ('x is ' , x)
         num1=float(instance1[x])
         num2=float(instance2[x])
         distance = distance + pow(num1-num2, 2)
   return math.sqrt(distance)
```

Questions:

1. What is the label for point if k=1?
2. What is the label for point if k=3?
3. Would the result be different if we used a different distance function like Hamming or Manhattan?

```python
#Add code with functions that implement Hamming and Manhattan distances and test
#what the label will be for k=1 and k=3 for all possibilities
#(6 total: 2 Euclidean, 2 Hamming and 2 Manhattan)

class distClass:
    def __init__(self, data, sample, length):
        self.data = data
        self.sample = sample
        self.length = length
        self.tag = data[-1]

    def get_euclideanDistance(self):
        distance = 0
        for i in range(self.length):
            num1 = float(self.data[i])
            num2 = float(self.sample[i])
            distance += pow(num1-num2, 2)
        return math.sqrt(distance)

    def get_manhattanDistance(self):
        dist = 0
        for i in range(self.length):
            num1 = float(self.data[i])
            num2 = float(self.sample[i])
            dist += abs(num1-num2)
        return dist

    def get_hammingDistance(self):
        distance = 0
        for x in range(self.length):
            if self.data[x] != self.sample[x]:
                distance += 1
        return distance


def get_list(data, sample, length):
    output = []
    for i in data:
        output.append(distClass(i, sample, length))
    return output

things = get_list(dataset, point, 3)

# Euclidean, k = 1
ordered = sorted(things, key=lambda thing: thing.get_euclideanDistance())
prediction = ordered[0].tag
print("Euclidean, k=1:", prediction)

# Euclidean, k = 3
tags = [ordered[0].tag, ordered[1].tag, ordered[2].tag]
prediction = max(tags, key=tags.count)
print("Euclidean, k=3:", prediction)

# Hamming, k = 1
ordered = sorted(things, key=lambda thing: thing.get_hammingDistance())
prediction = ordered[0].tag
print("Hamming, k=1:", prediction)

# Hamming, k = 3
tags = [ordered[0].tag, ordered[1].tag, ordered[2].tag]
prediction = max(tags, key=tags.count)
print("Hamming, k=3:", prediction)

# Manhattan, k = 1
ordered = sorted(things, key=lambda thing: thing.get_manhattanDistance())
prediction = ordered[0].tag
print("Manhattan, k=1:", prediction)

# Manhattan, k = 3
tags = [ordered[0].tag, ordered[1].tag, ordered[2].tag]
prediction = max(tags, key=tags.count)
print("Manhattan, k=3:", prediction)
```

Precision, Recall and F-measure:

Accuracy alone doesn't tell the whole story, so for each of the 9 results above you must also report the precision, recall, and f-measure.

Compute them separately for each tag in the dataset. For a given tag T:

Precision(T) = (number of test vectors correctly classified as T) / (number of test vectors your model classified as T)

Recall(T) = (number of test vectors correctly classified as T) / (number of test vectors whose true tag is T)

F-measure(T) = 2 * Precision(T) * Recall(T) / (Precision(T) + Recall(T))

Once you are done, each of your 9 results should include accuracy, precision, recall, and f-measure.

```python
# Measure each tag separately, treating that tag as positive.
def do_recall(results, test_data, tag):
    out = {}
    correct = float(sum(result["k1"] == sample[-1] == tag for result, sample in zip(results, test_data)))
    total = float(sum(t[-1] == tag for t in test_data))
    out["k1"] = correct / total if total else 0
    correct = float(sum(result["k7"] == sample[-1] == tag for result, sample in zip(results, test_data)))
    total = float(sum(t[-1] == tag for t in test_data))
    out["k7"] = correct / total if total else 0
    correct = float(sum(result["k15"] == sample[-1] == tag for result, sample in zip(results, test_data)))
    total = float(sum(t[-1] == tag for t in test_data))
    out["k15"] = correct / total if total else 0
    return out

def do_precision(results, test_data, tag):
    out = {}
    correct = float(sum(result["k1"] == sample[-1] == tag for result, sample in zip(results, test_data)))
    total = float(sum(t["k1"] == tag for t in results))
    out["k1"] = correct / total if total else 0
    correct = float(sum(result["k7"] == sample[-1] == tag for result, sample in zip(results, test_data)))
    total = float(sum(t["k7"] == tag for t in results))
    out["k7"] = correct / total if total else 0
    correct = float(sum(result["k15"] == sample[-1] == tag for result, sample in zip(results, test_data)))
    total = float(sum(t["k15"] == tag for t in results))
    out["k15"] = correct / total if total else 0
    return out

def do_f_measure(precision, recall):
    out = {}
    total = precision["k1"] + recall["k1"]
    out["k1"] = 2 * precision["k1"] * recall["k1"] / total if total else 0
    total = precision["k7"] + recall["k7"]
    out["k7"] = 2 * precision["k7"] * recall["k7"] / total if total else 0
    total = precision["k15"] + recall["k15"]
    out["k15"] = 2 * precision["k15"] * recall["k15"] / total if total else 0
    return out

# Euclidean
for tag in ["M", "F"]:
    print()
    print(f"Euclidean - tag {tag}")
    precision = do_precision(euclidean_results, test_data, tag)
    recall = do_recall(euclidean_results, test_data, tag)
    f_measure = do_f_measure(precision, recall)

    print(f"Euclidean, k=1, tag={tag}: precision = {precision['k1']:.2%}, recall = {recall['k1']:.2%}, F-measure = {f_measure['k1']:.2%}")
    print(f"Euclidean, k=7, tag={tag}: precision = {precision['k7']:.2%}, recall = {recall['k7']:.2%}, F-measure = {f_measure['k7']:.2%}")
    print(f"Euclidean, k=15, tag={tag}: precision = {precision['k15']:.2%}, recall = {recall['k15']:.2%}, F-measure = {f_measure['k15']:.2%}")

# Hamming
for tag in ["M", "F"]:
    print()
    print(f"Hamming - tag {tag}")
    precision = do_precision(hamming_results, test_data, tag)
    recall = do_recall(hamming_results, test_data, tag)
    f_measure = do_f_measure(precision, recall)

    print(f"Hamming, k=1, tag={tag}: precision = {precision['k1']:.2%}, recall = {recall['k1']:.2%}, F-measure = {f_measure['k1']:.2%}")
    print(f"Hamming, k=7, tag={tag}: precision = {precision['k7']:.2%}, recall = {recall['k7']:.2%}, F-measure = {f_measure['k7']:.2%}")
    print(f"Hamming, k=15, tag={tag}: precision = {precision['k15']:.2%}, recall = {recall['k15']:.2%}, F-measure = {f_measure['k15']:.2%}")

# Manhattan
for tag in ["M", "F"]:
    print()
    print(f"Manhattan - tag {tag}")
    precision = do_precision(manhattan_results, test_data, tag)
    recall = do_recall(manhattan_results, test_data, tag)
    f_measure = do_f_measure(precision, recall)

    print(f"Manhattan, k=1, tag={tag}: precision = {precision['k1']:.2%}, recall = {recall['k1']:.2%}, F-measure = {f_measure['k1']:.2%}")
    print(f"Manhattan, k=7, tag={tag}: precision = {precision['k7']:.2%}, recall = {recall['k7']:.2%}, F-measure = {f_measure['k7']:.2%}")
    print(f"Manhattan, k=15, tag={tag}: precision = {precision['k15']:.2%}, recall = {recall['k15']:.2%}, F-measure = {f_measure['k15']:.2%}")
```
